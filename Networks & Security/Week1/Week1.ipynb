```python
def find_largest_number(numbers):
    if not numbers:
        return None
    
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest

my_list = [3, 7, 2, 9, 4]
print(find_largest_number(my_list))  # Output: 9
```

```python
# Write a function that takes a string as a parameter and returns the number of vowels in the string.

def count_vowels(s):
    vowels = ('a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U')
    count = 0
    for char in s:
        if char in vowels:
            count += 1
    return count

input_string = "Hello, World!"
print(count_vowels(input_string))
```

```python
def check_password_strength(password):
    failures = []
    
    if len(password) < 8:
        failures.append("be at least 8 characters long")
        
    has_upper = False
    has_lower = False
    has_special = False
    
    for char in password:
        if char.isupper():
            has_upper = True
        elif char.islower():
            has_lower = True
        elif not char.isalnum():
            has_special = True

    if not has_upper:
        failures.append("contain at least one uppercase letter")
    if not has_lower:
        failures.append("contain at least one lowercase letter")
    if not has_special:
        failures.append("contain at least one special character")

    if not failures:
        print("Password is strong")
    else:
        if len(failures) == 1:
            missing_text = failures[0]
        elif len(failures) == 2:
            missing_text = f"{failures[0]} and {failures[1]}"
        else:
            missing_text = ", ".join(failures[:-1]) + f", and {failures[-1]}"
            
        print(f"Password must {missing_text}.")

password = "Pa$$word123"
check_password_strength(password)

password = "password"
check_password_strength(password)
```

```python
def caesar_cipher(message, key, mode):
    result = ""
    
    if mode == "decrypt":
        key = -key
        
    for char in message:
        if char.isupper():
            start = ord('A')
            shifted = (ord(char) - start + key) % 26 + start
            result += chr(shifted)
        elif char.islower():
            start = ord('a')
            shifted = (ord(char) - start + key) % 26 + start
            result += chr(shifted)
        else:
            result += char
            
    return result

user_message = input("Enter a message to encrypt: ")
key = 3

encrypted_message = caesar_cipher(user_message, key, "encrypt")
print("Encrypted:", encrypted_message)

decrypted_message = caesar_cipher(encrypted_message, key, "decrypt")
print("Decrypted back:", decrypted_message)
```

```python
import random

def play_hangman():
    words = ["python", "java", "javascript"]
    secret_word = random.choice(words)
    max_guesses = 6
    incorrect_guesses = 0
    guessed_letters = []

    while incorrect_guesses < max_guesses:
        # Display current word status with spaces
        display_word = ""
        for char in secret_word:
            if char in guessed_letters:
                display_word += char + " "
            else:
                display_word += "_ "
        print(display_word.strip())

        # Check if user has guessed all letters
        if "_" not in display_word:
            print(f'Congratulations, you guessed the word "{secret_word}"!')
            return

        guess = input("Guess a letter: ").lower()

        # Handle invalid or repeated input
        if len(guess) != 1 or not guess.isalpha():
            print("Please enter a single valid letter.")
            continue
        if guess in guessed_letters:
            print("You already guessed that letter.")
            continue

        guessed_letters.append(guess)

        if guess not in secret_word:
            incorrect_guesses += 1

    print(f'Game over! The word was "{secret_word}".')

play_hangman()
```
